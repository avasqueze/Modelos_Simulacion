# **PASOS EN UN BUEN ESTUDIO DE SIMULACIÓN**

> **Nota:** La programación del modelo es solo una parte del esfuerzo total para diseñar o analizar un sistema complejo mediante simulación. También se deben atender otras cuestiones como la aleatoriedad del sistema, la validación, el análisis estadístico de los datos de salida y la gestión del proyecto. Un estudio de simulación **no** es un proceso estrictamente secuencial; a medida que se avanza puede ser necesario **volver a pasos anteriores**.

---

## **Visión general del proceso (resumen)**

1. Formular el problema y planear el estudio
2. Recolectar datos y definir un modelo
3. Validar el **documento de supuestos**
4. Construir el programa y **verificarlo** (depuración)
5. Hacer **corridas piloto**
6. Validar el **modelo programado**
7. Diseñar los experimentos
8. Hacer **corridas de producción**
9. Analizar los **datos de salida**
10. Documentar, presentar y **usar** resultados

### **1. Formular el problema y planear el estudio**

**a.** El gerente expone el problema de interés.

* El problema puede **no** estar formulado correctamente o en términos cuantitativos.
* A menudo se requiere un proceso **iterativo**.

**b.** Realizar una o más reuniones de inicio (**kickoff**) con el gerente del proyecto, analistas de simulación y expertos en el dominio (SMEs). Discutir:

* Objetivos generales del estudio.
* Preguntas específicas a responder (necesarias para decidir el **nivel de detalle** del modelo).
* **Medidas de desempeño** para evaluar la eficacia de diferentes configuraciones del sistema.
* Alcance del modelo.
* Configuraciones del sistema a modelar (necesarias para decidir la **generalidad** del programa de simulación).
* **Cronograma** del estudio y recursos requeridos.

**c.** Seleccionar el **software** para el modelo.

---

### **2. Recolectar datos y definir un modelo**

**a.** Recolectar información sobre la **estructura** del sistema y los **procedimientos operativos**.

* Ninguna persona o documento por sí solos son suficientes.
* Algunas personas pueden tener información inexacta; **identificar** a los verdaderos **SMEs**.
* Los procedimientos operativos podrían no estar formalizados.

**b.** Recolectar datos (si es posible) para especificar **parámetros del modelo** y **distribuciones** de entrada.

**c.** Delimitar la información anterior y los datos en un **documento de supuestos** por escrito.

**d.** Recolectar datos (si es posible) sobre el **desempeño del sistema existente** (para validación en el paso 6).

**e.** Elegir el **nivel de detalle** del modelo (es un arte) según:

* Objetivos del proyecto.
* Medidas de desempeño.
* Disponibilidad de datos.
* Preocupaciones de **credibilidad**.
* Restricciones computacionales.
* Opiniones de los SMEs.
* Restricciones de tiempo y presupuesto.

**f.** No debe haber una **correspondencia uno a uno** entre cada elemento del modelo y su elemento homólogo en el sistema real.

**g.** Comenzar con un modelo **“simple”** y **enriquecerlo** según sea necesario. Modelar cada aspecto del sistema rara vez es necesario para tomar buenas decisiones, y puede causar **tiempos de ejecución excesivos**, incumplir plazos u **oscurecer** factores importantes.

**h.** Interactuar regularmente con el gerente (y personal clave del proyecto).

---

### **3. ¿El documento de supuestos es válido?**

**a.** Realizar una **revisión estructurada** del documento de supuestos con gerentes, analistas y SMEs. Esto:

* Ayuda a asegurar que los supuestos del modelo sean **correctos y completos**.
* **Promueve la interacción** entre los miembros del proyecto.
* **Fomenta la apropiación** del modelo.
* Se hace **antes** de programar, para evitar **reprogramaciones** significativas.

---

### **4. Construir el programa y verificar**

**a.** Programar el modelo en un **lenguaje** (p. ej., C/C++/Java) o en **software de simulación** (p. ej., Arena, ExtendSim, Flexsim, ProModel).

* Lenguajes: suelen ser conocidos, dan **mayor control**, bajo costo de adquisición y pueden lograr **menor tiempo** de ejecución.
* Software de simulación: **reduce** el tiempo de programación y el **costo** del proyecto.

**b.** **Verificar** (depurar) el programa de simulación.

---

### **5. Hacer corridas piloto**

**a.** Realizar **corridas piloto** con fines de validación en el paso 6.

---

### **6. ¿El modelo programado es válido?**

**a.** Si existe un sistema real, **comparar** las medidas de desempeño del modelo y del sistema real (datos del paso 2) para el sistema existente.

**b.** Independientemente de lo anterior, analistas y SMEs deben **revisar** los resultados del modelo para comprobar su **corrección**.

**c.** Usar **análisis de sensibilidad** para determinar qué factores del modelo impactan significativamente las medidas de desempeño y, por tanto, deben **modelarse con cuidado**.

---

### **7. Diseñar experimentos**

**a.** Especificar, para cada configuración de interés:

* **Duración** de cada corrida de simulación.
* **Período de calentamiento** (*warm-up*), si aplica.
* **Número de corridas independientes** con semillas aleatorias distintas (facilita construir **intervalos de confianza**).

---

### **8. Hacer corridas de producción**

**a.** Realizar varias corridas de producción antes de pasar al paso 9.

---

### **9. Analizar los datos de salida**

**a.** Dos objetivos principales:

* Determinar el **desempeño absoluto** de ciertas configuraciones del sistema.
* **Comparar** alternativas en **términos relativos**.

---

### **10. Documentar, presentar y usar resultados**

**a.** Documentar los **supuestos**, el **programa** y los **resultados** del estudio para uso actual y futuro.

**b.** Presentar los resultados del estudio.

* Utilizar **animación** para comunicar el modelo a gerentes y a quienes no conocen todos los detalles.
* Describir el proceso de **construcción** y **validación** del modelo para promover la **credibilidad**.
* Los resultados se usan en la **toma de decisiones** si son **válidos** y **creíbles**.

---

## **VENTAJAS, DESVENTAJAS Y ERRORES COMUNES DE LA SIMULACIÓN**

La simulación es un método ampliamente utilizado y cada vez más popular para estudiar sistemas complejos. A continuación se listan características y errores comunes.

### Posibles **ventajas** de la simulación

* La mayoría de los sistemas reales complejos con elementos **estocásticos** no pueden describirse con precisión mediante un modelo analítico evaluable; la simulación suele ser la **única** vía de investigación.
* Permite **estimar** el desempeño de un sistema existente bajo condiciones operativas proyectadas.
* **Comparar** diseños de sistemas propuestos (o políticas operativas) para ver cuál cumple mejor un requisito.
* Posibilita **mayor control** sobre las condiciones experimentales que al experimentar con el sistema real.
* Permite estudiar un sistema en **horizontes largos** en **tiempo comprimido**, o analizar su operación detallada en **tiempo expandido**.

### Posibles **desventajas**

* Cada corrida de un modelo estocástico produce solo **estimaciones** de las verdaderas características del modelo para un conjunto dado de entradas; se requieren **varias corridas independientes** por conjunto de parámetros. Por ello, la simulación es mejor para **comparar alternativas** que para **optimización**. Si existe un **modelo analítico válido**, suele ser preferible.
* Los modelos de simulación pueden ser **costosos** y **lentos** de desarrollar.
* El gran volumen de números o el impacto persuasivo de una **animación realista** puede llevar a **excesiva confianza** en resultados no justificados. Si el modelo **no** es una representación **válida**, los resultados —por muy impresionantes que parezcan— aportarán **poca** información útil.

### **Trampas** (errores comunes) que pueden arruinar un estudio

* No definir **claramente** los **objetivos** al inicio.
* No involucrar a **todo el equipo** desde el comienzo.
* Elegir un **nivel de detalle** inapropiado.
* Falta de **comunicación** con la gerencia durante el estudio.
* **Malentendidos** sobre lo que es la simulación por parte de la gerencia.
* Tratar la simulación como si fuera principalmente un ejercicio de **programación**.
* Falta de personal con **metodología de simulación** y **estadística** en el equipo.
* No recolectar **buenos datos** del sistema.
* Seleccionar **software inapropiado**.
* Usar paquetes con **macros complejas** poco documentadas que no implementan la lógica deseada.
* Creer que paquetes “fáciles de usar” requieren **baja** competencia técnica.
* **Mal uso** de la animación.
* No considerar correctamente las **fuentes de aleatoriedad** del sistema real.
* Usar **distribuciones arbitrarias** (normal, uniforme, triangular) como entradas sin justificación.
* Analizar la **salida de una sola corrida** con fórmulas que **asumen independencia**.
* Hacer **una sola réplica** y tratar sus estadísticas como “la verdad”.
* No incluir **período de calentamiento** cuando interesa el comportamiento en **régimen permanente**.
* Comparar diseños alternativos con **una sola réplica** por diseño.
* Usar **medidas de desempeño** equivocadas.

> **Recomendación:** según el contexto, pueden ser útiles **ambos** enfoques (analítico y de simulación). La simulación puede **verificar** supuestos de un modelo analítico; el modelo analítico puede **sugerir** alternativas a investigar mediante simulación.
