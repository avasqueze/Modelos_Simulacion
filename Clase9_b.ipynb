## Ecuaciones en diferencias

Los modelos de población del capítulo anterior y de este son lo suficientemente simples
como para que realmente no necesitáramos ejecutar simulaciones. Podríamos haberlos
resuelto matemáticamente. Por ejemplo, escribimos el modelo de crecimiento constante así:

$$
resultado[t+1] = resultado[t] + CreciemientoAnual
$$

En notación matemática, escribiríamos el mismo modelo así:

$$
x_{n+1} = x_n + c
$$ 

donde $x_n$ es la población durante el año $n$, 
$x_{n+1}$ es la población durante el año $n+1$,
y $c$ es el crecimiento anual constante.
Esta manera de representar el modelo, donde la población futura es una función de la población actual, es una *ecuación en diferencias*; ver
<https://es.wikipedia.org/wiki/Ecuaci%C3%B3n_en_diferencia_lineal>.

En el caso de crecimiento constante podemos ver que $x_1 = x_0 + c$, y
$x_2 = x_1 + c$. Combinando estos, obtenemos que $x_2 = x_0 + 2c$, luego
$x_3 = x_0 + 3c$, y podemos ver que en general

$$
x_n = x_0 + nc
$$ 

Así que si queremos saber $x_{100}$ y no nos importan los otros valores, podemos calcularlo con una multiplicación y una suma.

También podemos escribir el modelo proporcional como una ecuación en diferencias:

$$
x_{n+1} = x_n + \alpha x_n
$$ 

O de forma más convencional:

$$
x_{n+1} = x_n (1 + \alpha)
$$ 

Ahora podemos ver que $x_1 = x_0 (1 + \alpha)$, y $x_2 = x_0 (1 + \alpha)^2$, y en general

$$
x_n = x_0 (1 + \alpha)^n
$$

Una sucesión con esta forma funcional se llama *progresión geométrica*. Cuando $\alpha$ es positivo, el factor
$1+\alpha$ es mayor que 1, por lo que los elementos de la sucesión crecen
sin límite.

Finalmente, podemos escribir el modelo cuadrático así:

$$
x_{n+1} = x_n + \alpha x_n + \beta x_n^2
$$ 

o con la parametrización más convencional:

$$
x_{n+1} = x_n + r x_n (1 - x_n / K)
$$ 

No existe una solución analítica para esta ecuación, pero podemos aproximar el modelo con una ecuación diferencial y resolverla, que es lo que haremos en la siguiente sección.

## Ecuaciones diferenciales

Volviendo al modelo de crecimiento constante 

$$
x_{n+1} = x_n + c
$$ 

Si definimos $\Delta x$ como el cambio en $x$ de un paso de tiempo al siguiente, podemos escribir: 

$$
\Delta x = x_{n+1} - x_n = c
$$ 

Si definimos
$\Delta t$ como el paso de tiempo, que en el ejemplo es un año, podemos
escribir la tasa de cambio por unidad de tiempo así:

$$
\frac{\Delta x}{\Delta t} = c
$$ 

Este es un modelo *discreto*,
lo que significa que el tiempo solo está definido en valores enteros de $n$ y no entre ellos.
Pero en la realidad, las personas nacen y mueren todo el tiempo, no una vez al año,
así que podría ser más realista usar un modelo *continuo*, lo que significa
que el tiempo está definido para todos los valores de $t$, no solo enteros. 

En un modelo continuo, escribimos la tasa de cambio en la
forma de una derivada: 

$$
\frac{dx}{dt} = c
$$ 

Esta forma de representar el modelo es una *ecuación diferencial*, que es una ecuación que involucra al menos una derivada.
Para resolver esta ecuación, multiplicamos ambos lados por $dt$: 

$$
dx = c dt
$$ 

Y luego integramos ambos lados: 

$$
x(t) = c t + x_0
$$

De forma similar, podemos escribir el modelo de crecimiento proporcional así:

$$
\frac{\Delta x}{\Delta t} = \alpha x
$$ 

Y como una ecuación diferencial
de esta manera: 

$$
\frac{dx}{dt} = \alpha x
$$ 

Si multiplicamos ambos lados por
$dt$ y dividimos por $x$, obtenemos 

$$
\frac{1}{x}~dx = \alpha~dt
$$ 

Ahora
integramos ambos lados, obteniendo: 

$$
\ln x = \alpha t + K
$$ 

donde $\ln$ es el logaritmo natural y $K$ es la constante de integración.

Al exponentiar ambos lados, tenemos

$$
\exp(\ln(x)) = \exp(\alpha t + K)
$$ 

La función exponencial puede escribirse como $\exp(x)$ o $e^x$. En este
libro uso la primera forma porque se parece al código de Python.
Podemos reescribir la ecuación anterior como

$$
x = \exp(\alpha t) \exp(K)
$$ 

Puesto que $K$ es una constante arbitraria,
$\exp(K)$ también es una constante arbitraria, así que podemos escribir

$$
x = C \exp(\alpha t)
$$ 

donde $C = \exp(K)$. Hay muchas soluciones para esta ecuación diferencial;
la solución que queremos es la que tiene el valor $x_0$ cuando $t=0$.

Cuando $t=0$, $x(t) = C$, así que $C = x_0$ y la solución que queremos es

$$
x(t) = x_0 \exp(\alpha t)
$$ 

## Análisis y simulación

Una vez que has diseñado un modelo, generalmente hay dos maneras de estudiarlo:
simulación y análisis. Por *análisis* me refiero al cálculo simbólico; es decir,
manipulación simbólica usando notación matemática.

El análisis y la simulación tienen **capacidades y limitaciones** diferentes.
La simulación suele ser más versátil; es fácil agregar y quitar componentes,
cambiar mecanismos y probar muchas versiones de un modelo, como hemos hecho
en los ejemplos anteriores.

Pero hay varias cosas que podemos hacer con el análisis que son más difíciles
o imposibles con simulaciones:

-   Con el análisis podemos **a veces** calcular, de manera **exacta y eficiente**, un
    valor que con simulación solo podríamos aproximar, y de forma menos eficiente.
    Por ejemplo, en el modelo cuadrático graficamos el crecimiento neto como
    función de la población y vimos que cruza por cero cuando la población está
    cerca de 14 mil millones.  Podríamos estimar el punto de cruce usando un
    algoritmo de búsqueda numérica (más sobre eso después). Pero con un poco de
    álgebra, derivamos el resultado general de que $K=-\alpha/\beta$.

-   El análisis a veces proporciona **"atajos computacionales"**, es decir, la
    capacidad de saltar hacia adelante en el tiempo para calcular el estado de un sistema
    muchos pasos de tiempo en el futuro sin calcular los estados intermedios.

-   Podemos usar análisis para enunciar y demostrar **generalizaciones** sobre los modelos;
    por ejemplo, podríamos demostrar que ciertos resultados siempre o
    nunca ocurren. Con simulaciones, podemos mostrar ejemplos y a veces
    encontrar contraejemplos, pero es difícil escribir pruebas formales.

-   El análisis puede aportar **comprensión** a los modelos y a los sistemas que
    describen; por ejemplo, a veces podemos identificar
    formas cualitativamente diferentes en que el sistema puede comportarse y **parámetros clave** que controlan
    esos comportamientos.

La notación matemática es un lenguaje
diseñado para expresar ideas computacionales; la programación es otro lenguaje
diseñado específicamente para representar ideas computacionales y ejecutar programas.

Cada uno de estos lenguajes es bueno para los fines para los que fue diseñado;
ninguno es superior en general. Uno de los objetivos del curso es mostrar
cómo pueden usarse juntos.

## Análisis con WolframAlpha

Hasta hace poco, la mayor parte del análisis se hacía frotando grafito sobre pulpa
de madera, un proceso laborioso y propenso a errores. Una alternativa útil es la
computación simbólica. Si has usado servicios como WolframAlpha, has usado
computación simbólica.

Por ejemplo, si vas a <https://www.wolframalpha.com/> e introduces

$$
df(t) / dt = \alpha f(t)
$$

WolframAlpha infiere que `f(t)` es una función de `t` y que $\alpha$ es un
parámetro; clasifica la consulta como una "ecuación diferencial ordinaria lineal de primer orden"
y reporta la solución general:

Si agregas una segunda ecuación para especificar la condición inicial:

$$
df(t) / dt = alpha f(t),  f(0) = p_0
$$

WolframAlpha reporta la solución particular:

$$
f(t) = p_0 \exp(\alpha t)
$$

WolframAlpha está basado en Mathematica, un potente lenguaje de programación
diseñado específicamente para computación simbólica.

## Análisis con SymPy

Python tiene una biblioteca llamada SymPy que proporciona herramientas de
computación simbólica similares a Mathematica. No son tan fáciles de usar como
WolframAlpha, pero tienen otras ventajas.

Para usarla, definiremos objetos `Symbol` que representan nombres de variables y funciones.
La función `symbols` toma una cadena y devuelve objetos `Symbol`.

```python
from sympy import symbols

t = symbols('t')
```

Ahora, cuando usemos `t`, Python lo tratará como un nombre de variable en lugar de un número específico. 
Por ejemplo, si usamos `t` como parte de una expresión, así:

```python
expr = t + 1
expr
```

```python
expr2 = t*t
expr2
```

Python no intenta realizar una suma numérica; más bien, crea un
nuevo `Symbol` que representa la suma de `t` y `1`. Podemos evaluar la
suma usando `subs`, que sustituye un valor por un símbolo. Este ejemplo
sustituye 2 por `t`:

```python
expr.subs(t, 2)
```

Las funciones en SymPy están representadas por un tipo especial de `Symbol`:

```python
from sympy import Function

f = Function('f')
f
```

Ahora, si escribimos `f(t)`, obtenemos un objeto que representa la evaluación de una función, $f$, en un valor, $t$.

```python
f(t)
```

Pero, de nuevo, SymPy no intenta
evaluarlo realmente.

## Ecuaciones diferenciales en SymPy

SymPy proporciona una función, `diff`, que puede derivar una función. Podemos aplicarla a `f(t)` así:

```python
from sympy import diff

dfdt = diff(f(t), t)
dfdt
```

El resultado es un `Symbol` que representa la derivada de `f` con
respecto a `t`. Pero, de nuevo, SymPy no intenta calcular la derivada
todavía.

Para representar una ecuación diferencial, usamos `Eq`:

```python
from sympy import Eq

alpha = symbols('alpha')
eq1 = Eq(dfdt, alpha*f(t))
eq1
```

El resultado es un objeto que representa una ecuación.  Ahora
podemos usar `dsolve` para resolver esta ecuación diferencial:

```python
from sympy import dsolve

solution_eq = dsolve(eq1)
solution_eq
```

El resultado es la *solución general*, que aún contiene una constante no especificada, $C_1$.
Para obtener la *solución particular* donde $f(0) = p_0$, sustituimos $p_0$ por `C1`. 
Primero, tenemos que indicarle a Python que `C1` es un símbolo.

```python
C1 = symbols('C1')
```

Ahora podemos sustituir el valor de $p_0$ por `C1`.
Por ejemplo, si $p_0$ es 1000:

```python
particular = solution_eq.subs(C1, 1000)
particular
```

Cuando $t=0$, el valor de $f(0)$ es $p_0$, lo que confirma que esta es la solución que queremos.

```python
particular.subs(t, 0)
```

## Resolver el modelo de crecimiento cuadrático

Para resolver la curva de crecimiento cuadrático, usaremos la parametrización $r, K$, así que necesitaremos dos símbolos más:

```python
r, K = symbols('r K')
```

Ahora podemos escribir la ecuación diferencial.

```python
eq2 = Eq(diff(f(t), t), r * f(t) * (1 - f(t)/K))
eq2
```

Y resolverla.

```python
solution_eq = dsolve(eq2)
solution_eq
```

El resultado, `solution_eq`, contiene `rhs`, que es el lado derecho de la solución.

```python
general = solution_eq.rhs
general
```

Podemos evaluar el lado derecho en $t=0$

```python
at_0 = general.subs(t, 0)
at_0
```

Ahora queremos encontrar el valor de `C1` que hace que `f(0) = p_0`.

Así que crearemos la ecuación `at_0 = p_0` y resolveremos para `C1`.  Como esto es una ecuación algebraica, no una ecuación diferencial, usamos `solve`, no `dsolve`.

```python
from sympy import solve

p_0 = symbols('p_0')
solutions = solve(Eq(at_0, p_0), C1)
```

El resultado de `solve` es una lista de soluciones.  

```python
type(solutions), len(solutions)
solutions
```

En este caso, solo hay una solución, pero aun así obtenemos una lista, por lo que
tenemos que usar el operador de corchetes, `[0]`, para seleccionar la primera.

```python
value_of_C1 = solutions[0]
value_of_C1
```

Ahora, en la solución general, queremos reemplazar `C1` con el valor de `C1` que acabamos de averiguar.

```python
particular = general.subs(C1, value_of_C1)
particular
```

El resultado es complicado, pero SymPy proporciona una función que intenta simplificarlo.

```python
simpler = particular.simplify()
simpler
```

Esta función se llama *curva de crecimiento logístico*. En el contexto de los modelos de crecimiento, la
función logística suele escribirse así:

$$
f(t) = \frac{K}{1 + A \exp(-rt)}
$$ 

donde $A = (K - p_0) / p_0$.

Podemos usar SymPy para confirmar que estas dos formas son equivalentes. Primero representamos la versión alternativa de la función logística:

```python
A = (K - p_0) / p_0
A
```

```python
from sympy import exp

logistic = K / (1 + A * exp(-r*t))
logistic
```

Para ver si dos expresiones son equivalentes, podemos comprobar si su diferencia se simplifica a 0.

```python
check = particular - logistic
check
```

```python
check.simplify()
```

Esta prueba solo funciona en un sentido: si SymPy dice que la diferencia se reduce a 0, entonces las expresiones son
definitivamente equivalentes (y no solo numéricamente cercanas).

Pero si SymPy no puede encontrar una manera de simplificar el resultado a 0, eso no prueba que sean diferentes; puede ser que SymPy simplemente no encuentre la manera de demostrarlo; de hecho, no existe un algoritmo que pueda resolverlo de forma general.

Si usas SymPy para calcular una expresión, y luego quieres evaluarla en Python, SymPy
proporciona una función llamada `pycode` que genera código de Python:

```python
from sympy.printing.pycode import pycode

pycode(check)
```

Finalmente, el modelo de crecimiento cuadrático se denomina *crecimiento logístico* porque la solución es una función logística:

$$
f(t) = \frac{K}{1 + A \exp(-rt)}
$$ 

donde $A = (K - p_0) / p_0$.

Con esto, terminamos de modelar el crecimiento de la población mundial.
